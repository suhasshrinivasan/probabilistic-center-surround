```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import pickle

seed = 42
rng = np.random.default_rng(seed)
```

```python
with open("prior_samples.pkl", "rb") as f:
    samples = pickle.load(f)
```

```python
samples.keys()
```

```python
fig, axs = plt.subplots(10, 10, figsize=(10, 10))
for idx, ax in enumerate(axs.ravel()):
    ax.imshow(samples["I"][idx], cmap="gray")
    ax.axis("off")
```

```python
fig, axs = plt.subplots(10, 10, figsize=(10, 10))
for idx, ax in enumerate(axs.ravel()):
    ax.imshow(samples["I_mu"][idx].reshape(36, 36), cmap="gray")
    ax.axis("off")
```

```python
x_samples = samples["X"]
x_samples.shape
```

```python
x_correlation = np.corrcoef(x_samples.T)
x_correlation.shape
```

```python
fig, ax = plt.subplots(dpi=200)
im = ax.imshow(x_correlation, cmap="seismic")
fig.colorbar(im, ax=ax)
```

```python
center_x_samples = x_samples[:, 160:199]
center_x_samples.shape
```

```python
center_x_correlation = np.corrcoef(center_x_samples.T)
```

```python
fig, ax = plt.subplots(dpi=200)
im = ax.imshow(center_x_correlation, cmap="seismic", vmin=-1, vmax=1)
fig.colorbar(im, ax=ax)
```
