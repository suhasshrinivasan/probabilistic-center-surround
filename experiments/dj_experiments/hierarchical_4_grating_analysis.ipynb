```python
print(optimal_df["completing_image_x_samples"].values[0].shape)
completing_samples = optimal_df["completing_image_x_samples"].values[0]
mei_samples = optimal_df["mei_x_samples"].values[0]
disrupting_samples = optimal_df["disrupting_image_1_x_samples"].values[0]
```

```python
fig, axs = plt.subplots(nrows=6, ncols=6, dpi=300, sharex=True, sharey=True)
for idx, ax in enumerate(axs.flatten()):
    sns.histplot(
        mei_samples[..., idx].flatten(),
        ax=ax,
        stat="density",
        color="blue",
        element="step",
        alpha=0.3,
        label="MEI"
    )
    sns.histplot(
        completing_samples[..., idx].flatten(),
        ax=ax,
        stat="density",
        color="green",
        element="step",
        alpha=0.3,
        label="completing"
    )
    sns.histplot(
        disrupting_samples[..., idx].flatten(),
        ax=ax,
        stat="density",
        color="red",
        element="step",
        alpha=0.3,
        label="disrupting"
    )
    # write the id of the neuron in the top left corner
    ax.text(0.8, 0.8, idx, transform=ax.transAxes, fontsize=6)
# figure legend
handles, labels = ax.get_legend_handles_labels()
fig.legend(handles, labels, loc='upper center')
```

```python
import numpy as np
fig, ax = plt.subplots(dpi=300)
neuron_id = 16
sns.histplot(
    mei_samples[..., neuron_id].flatten(),
    ax=ax,
    stat="density",
    color="blue",
    element="step",
    alpha=0.3,
    label="MEI"
)
sns.histplot(
    completing_samples[..., neuron_id].flatten(),
    ax=ax,
    stat="density",
    color="green",
    element="step",
    alpha=0.3,
    label="completing"
)
sns.histplot(
    disrupting_samples[..., neuron_id].flatten(),
    ax=ax,
    stat="density",
    color="red",
    element="step",
    alpha=0.3,
    label="disrupting"
)

MEI_mean = np.mean(mei_samples[..., neuron_id].flatten())
completing_mean = np.mean(completing_samples[..., neuron_id].flatten())
disrupting_mean = np.mean(disrupting_samples[..., neuron_id].flatten())
completing_perc_change = (completing_mean - MEI_mean) / MEI_mean
disrupting_perc_change = (disrupting_mean - MEI_mean) / MEI_mean

ax.text(0.7, 0.9, f"$\mu$={MEI_mean:.4f}", size=10, ha="left", va="bottom", transform=ax.transAxes, color='blue')
ax.text(0.7, 0.8, f"$\mu$={completing_mean:.4f}, $\Delta$={completing_perc_change*100:.1f}%", size=10, ha="left", va="bottom", transform=ax.transAxes, color='green')
ax.text(0.7, 0.7, f"$\mu$={disrupting_mean:.4f}, $\Delta$={disrupting_perc_change*100:.1f}%", size=10, ha="left", va="bottom", transform=ax.transAxes, color='red')

ax.axvline(MEI_mean, color="blue", linestyle="--", linewidth=0.5, label="MEI $\mu$")
ax.axvline(completing_mean, color="green", linestyle="--", linewidth=0.5, label="completing $\mu$")
ax.axvline(disrupting_mean, color="red", linestyle="--", linewidth=0.5, label="disrupting $\mu$")

ax.legend(loc="upper left")
```
