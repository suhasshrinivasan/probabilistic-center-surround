```python
from sklearn.decomposition import FastICA
import numpy as np
from numpy.random import default_rng
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
import torch
import torch.distributions as D
from torchvision import transforms
from torch.utils.data import DataLoader
from torchvision import transforms
from torchvision.datasets import DatasetFolder
import os
from os import listdir
from os.path import isfile, join

seed = 42
rng = default_rng(seed=seed)
torch.manual_seed(seed=seed)
EPSILON = 1e-12
mpl.style.use("seaborn")
```

## Load data

```python
data_folder = "/src/project/data/"
dataset = DatasetFolder(
    root=data_folder,
    loader=lambda path: torch.from_numpy(np.load(path)),
    extensions=".npy",
    transform=transforms.RandomCrop(12)
)
```

```python
images, labels = next(iter(DataLoader(dataset, batch_size=dataset.__len__())))
```

```python
stimuli = images.flatten(start_dim=1)
```

```python
transformer_under = FastICA(n_components=40, random_state=seed)
z_under = transformer_under.fit_transform(stimuli)
```

```python
transformer = FastICA(n_components=144, random_state=seed)
z = transformer.fit_transform(stimuli)
```
