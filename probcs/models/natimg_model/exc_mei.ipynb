```python
import cv2
```

```python
def resize(images, size):
    resized = []
    for image in images:
        resized.append(cv2.resize(image, size))
    return np.array(resized)
```

```python
def crop_and_resize(images, size=(36, 36), anti_aliasing=False):
    final_images = []
    for image in images:
        h, w = image.shape
        cropped = image[:, int((w - h) / 2) : -int((w - h) / 2)]
        resized = resize(cropped, size, anti_aliasing=anti_aliasing)
        final_images.append(resized)
    return np.array(final_images)
```
