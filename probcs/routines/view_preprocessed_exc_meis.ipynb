```python
import numpy as np
import pickle
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
basepath = Path("/src/project/data/experiment")
cropped_path = basepath / "exc_images_preprocessed.npy"
cropped_30_path = basepath / "exc_images_preprocessed_24x24.npy"
```

```python
cropped = np.load(cropped_path)
cropped_30 = np.load(cropped_30_path)
```

```python
cropped[0].shape, cropped_30[0].shape
```

```python
fig, axs = plt.subplots(10, 2, figsize=(2 * 5, 10 * 5))
for idx, (crop, center_crop) in enumerate(zip(cropped, cropped_30)):
    if idx > 9:
        break
    axs[idx, 0].imshow(crop, cmap="gray")
    axs[idx, 0].axis("off")
    axs[idx, 1].imshow(center_crop, cmap="gray")
    axs[idx, 1].axis("off")
```
